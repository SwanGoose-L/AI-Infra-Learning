````python
def get_stats(ids):
    counts = {}
    for pair in zip(ids, ids[1:]): # 用符合 Python 风格的写法遍历相邻元素对
        counts[pair] = counts.get(pair, 0) + 1
    return counts

stats = get_stats(tokens)
# print(stats)
print(sorted(((v,k) for k,v in stats.items()), reverse=True))
````

````python
top_pair = max(stats, key=stats.get)
top_pair
````

````python
def merge(ids, pair, idx):
  # 在整数列表 ids 中，将所有连续出现的 pair 替换为新 token 的编号 idx
  newids = []
  i = 0
  while i < len(ids):
    # 如果当前不是最后一个位置，并且相邻元素与 pair 匹配，就进行替换
    if i < len(ids) - 1 and ids[i] == pair[0] and ids[i+1] == pair[1]:
      newids.append(idx)
      i += 2
    else:
      newids.append(ids[i])
      i += 1
  return newids

print(merge([5, 6, 6, 7, 9, 1], (6, 7), 99))

tokens2 = merge(tokens, top_pair, 256)
print(tokens2)
print("length:", len(tokens2))
````

````python
def get_stats(ids):
    counts = {}
    for pair in zip(ids, ids[1:]):
        counts[pair] = counts.get(pair, 0) + 1
    return counts

def merge(ids, pair, idx):
  newids = []
  i = 0
  while i < len(ids):
    if i < len(ids) - 1 and ids[i] == pair[0] and ids[i+1] == pair[1]:
      newids.append(idx)
      i += 2
    else:
      newids.append(ids[i])
      i += 1
  return newids

# ---
vocab_size = 276 # 期望的最终词表大小
num_merges = vocab_size - 256
ids = list(tokens) # 创建副本，避免修改原列表

merges = {} # (int, int) -> int
for i in range(num_merges):
  stats = get_stats(ids)
  pair = max(stats, key=stats.get)
  idx = 256 + i
  print(f"merging {pair} into a new token {idx}")
  ids = merge(ids, pair, idx)
  merges[pair] = idx
````

### 解码

给定一个取值范围为 [0, vocab_size] 的整数序列，它对应什么文本？


https://en.wikipedia.org/wiki/UTF-8


````python
vocab = {idx: bytes([idx]) for idx in range(256)}
# - merges.items() 需要按训练时插入合并规则的顺序遍历，Python 3.7起可以保证
for (p0, p1), idx in merges.items():
    vocab[idx] = vocab[p0] + vocab[p1]

def decode(ids):
  # 给定 ids（整数列表），返回 Python 字符串
  tokens = b"".join(vocab[idx] for idx in ids)
  text = tokens.decode("utf-8", errors="replace")
  return text

print(decode([97]))
print(decode([128]))
````

反过来：给定一个字符串，它对应哪些 token？

````python
def encode(text):
  # 给定字符串，返回整数列表（即 token 编号）
  tokens = list(text.encode("utf-8"))
  while len(tokens) >= 2:
    stats = get_stats(tokens)
    pair = min(stats, key=lambda p: merges.get(p, float("inf")))
    if pair not in merges:
      break # 已经没有可以继续合并的元素对
    idx = merges[pair]
    tokens = merge(tokens, pair, idx)
  return tokens

print(encode(""))
````

````python
# 使用该文本训练 sentencepiece 模型
# 这里的设置尽可能与训练 Llama 2 时采用的设置保持一致
import os

options = dict(
  # 输入设置
  input="toy.txt",
  input_format="text",
  # 输出设置
  model_prefix="tok400", # 输出文件名的前缀
  # 算法设置
  # BPE 算法
  model_type="bpe",
  vocab_size=400,
  # 归一化
  normalization_rule_name="identity", # 关闭归一化
  remove_extra_whitespaces=False,
  input_sentence_size=200000000, # 训练句子的最大数量
  max_sentence_length=4192, # 每个句子的最大字节数
  seed_sentencepiece_size=1000000,
  shuffle_input_sentence=True,
  # 罕见词处理
  character_coverage=0.99995,
  # 字节回退
  byte_fallback=True,
  # 合并规则
  split_digits=True,
  split_by_unicode_script=True,
  split_by_whitespace=True,
  split_by_number=True,
  max_sentencepiece_length=16,
  add_dummy_prefix=True,
  allow_whitespace_only_pieces=True,
  # 特殊 token
  unk_id=0, # UNK token 必须存在
  bos_id=1, # 其他特殊 token 为可选项，设为 -1 可禁用
  eos_id=2,
  pad_id=-1,
  # 系统设置
  num_threads=os.cpu_count(), # 使用几乎全部系统资源
)

spm.SentencePieceTrainer.train(**options)
````
